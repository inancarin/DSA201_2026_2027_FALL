```python
y_values = []
x_values = []
for i in range(1, 101):
    y_values.append(i**2)
    x_values.append(i)

plt.plot(x_values, y_values, label="y=x^2")
plt.legend()
#plt.xlim(2, 5)
#plt.ylim(4, 25)
plt.show()
```

```python
for k in range(1, 5): # k -> 1,2,3,4 
    x_values = []
    y_values = []
    for i in range(1, 6):
        x_values.append(i)
        y_values.append(i*k)
    plt.plot(x_values, y_values, label="y=" + str(k) + "x")
plt.legend()
plt.show()
```

```python
f = open("iris.txt")

allLines = f.readlines()

f.close()

x = []
y = []
color_codes = []
d = {}
for line in allLines[1:]:
    line = line.strip()
    info = line.split(",")
    sep_len = float(info[1])
    sep_wid = float(info[2])
    species = info[-1]
    if species not in d:
        d[species] = len(d)

    color_codes.append(d[species])
    x.append(sep_len)
    y.append(sep_wid)

plt.scatter(x, y, c=color_codes)
plt.show()
```

```python
f = open("iris.txt")

allLines = f.readlines()

f.close()

d = {}
for line in allLines[1:]:
    line = line.strip()
    info = line.split(",")
    sep_len = float(info[1])
    sep_wid = float(info[2])
    species = info[-1]

    if species not in d:
        d[species] = {"x": [], 
                      "y": []}
    d[species]["x"].append(sep_len)
    d[species]["y"].append(sep_wid)


for k, v in d.items():
    plt.scatter(v["x"], v["y"], label=k)
plt.legend()
plt.show()
```
