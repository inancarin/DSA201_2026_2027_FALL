```python
for k in range(1, 5): # k -> 1,2,3,4 
    x_values = []
    y_values = []
    for i in range(1, 6):
        x_values.append(i)
        y_values.append(i*k)
    plt.plot(x_values, y_values, label="y=" + str(k) + "x")
plt.legend()
plt.show()
```

```python
f = open("iris.txt")

allLines = f.readlines()

f.close()

x = []
y = []
color_codes = []
d = {}
for line in allLines[1:]:
    line = line.strip()
    info = line.split(",")
    sep_len = float(info[1])
    sep_wid = float(info[2])
    species = info[-1]
    if species not in d:
        d[species] = len(d)

    color_codes.append(d[species])
    x.append(sep_len)
    y.append(sep_wid)

plt.scatter(x, y, c=color_codes)
plt.show()
```

```python
f = open("iris.txt")

allLines = f.readlines()

f.close()

d = {}
for line in allLines[1:]:
    line = line.strip()
    info = line.split(",")
    sep_len = float(info[1])
    sep_wid = float(info[2])
    species = info[-1]

    if species not in d:
        d[species] = {"x": [], 
                      "y": []}
    d[species]["x"].append(sep_len)
    d[species]["y"].append(sep_wid)


for k, v in d.items():
    plt.scatter(v["x"], v["y"], label=k)
plt.legend()
plt.show()
```

```python
f = open("GOOGL.txt")

allLines = f.readlines()

d = dict() # d = {}
for line in allLines[1:]:
    info = line.split(",")
    date = info[0]
    close = float(info[4])
    year = int(date[:4])
    if 2004 < year < 2022:
        if year not in d:
            d[year] = []
        d[year].append(close)

f.close()

"""
plt.plot(range(1, 4), [101.456459,97.347343,96.851852], label=2005)
plt.plot(range(1, 4), [102.456459,96.347343,96.851852], label=2006)
plt.plot(range(1, 4), [105.456459,99.347343,96.851852], label=2007)
...
plt.plot(range(1, 5), [110.456459,97.347343,96.851852, 99], label=2021)
plt.show()
"""

plt.figure(figsize=(15, 8))
for year, stock_prices in d.items():
    plt.plot(range(1, len(stock_prices)+1), stock_prices, label=year)
plt.xlabel("Day of the Year")
plt.ylabel("Close Value")
plt.legend()
plt.show()
```
