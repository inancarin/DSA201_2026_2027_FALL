```python
df = pd.read_csv("http://bit.ly/kaggletrain")

for i in range(1000):
    rand_indices = df.sample(5, replace=True).index
    if len(rand_indices.unique()) != 5:
        print(rand_indices)
```

```python
for i in range(1000):
    rand_indices = df.sample(5, replace=False).index
    if len(rand_indices.unique()) != 5:
        print(rand_indices)
```

```python
thr = 90
n_iter = 1000
averages = []
for i in range(n_iter):
    temp_avg = df["Fare"].sample(100, replace=True).mean()
    averages.append(temp_avg)

averages.sort()
idx = int(n_iter * (100 - thr) / 2 / 100)


thr1 = averages[idx-1]
thr2 = averages[-idx]
#print(thr1, thr2)

sns.displot(averages)
# point1 = (thr1, 0)
# point2 = (thr1, 125)
plt.plot([thr1, thr1], [0, 125], color="red", linestyle="dashed", linewidth=3, label=format(thr1, ".2f"))
plt.plot([thr2, thr2], [0, 125], color="red", linestyle="dashed", linewidth=3, label=format(thr2, ".2f"))
plt.title("A sample of 1000 means of 100 with" + str(thr) + "percent of confidence intervals")
plt.legend()
plt.show()
```
