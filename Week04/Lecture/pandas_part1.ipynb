```python
durations = []
for dur in movies.duration:
    if dur > 180:
        durations.append(True)
    else:
        durations.append(False)

durations[:10]
```

```python
movies[durations]
```

```python
movies[movies["duration"] > 180]
```

```python
movies.head()
```

```python
# find the movies whose length is more than 199 minutes

movies[movies["duration"] > 199]
```

```python
"""
and -> &
or -> |
"""
```

```python
# find the movies whose length is more than 199 minutes and genre is Drama

movies[(movies["duration"]  > 199) & (movies["genre"] == "Drama")]
```

```python
c1 = movies["duration"]  > 199
c2 = movies["genre"] == "Drama"

movies[c1 & c2]
```

```python
# find the movies whose length is more than 199 minutes and genre is Drama or Crime

movies[(movies["duration"]  > 199) & ((movies["genre"] == "Drama") | (movies["genre"] == "Crime"))]
```

```python
movies[(movies["duration"]  > 199) & (movies["genre"].isin(["Drama", "Crime"]))]
```

```python
# calculate the avg price of "Chicken Bowl" items

df = orders[orders["item_name"] == "Chicken Bowl"]
df["item_price"].str.replace("$", "").astype(float).mean()
```

```python
# calculate the avg price of items whose name contains "chicken" or "ChiCKen" etc


df = orders[orders["item_name"].str.lower().str.contains("chicken")]
df["item_price"].str.replace("$", "").astype(float).mean()
```

```python
drinks = pd.read_csv("http://bit.ly/drinksbycountry")
drinks.head()
```

```python
# let's find European country that consumes most beer
res = drinks[drinks["continent"] == "Europe"].sort_values("beer_servings", ascending=False)["country"].values[0]
res
```

```python
# let's find avg beer cons in Europe

drinks[drinks["continent"] == "Europe"]["beer_servings"].mean()
```

```python
for cont in drinks["continent"].unique():
    res = drinks[drinks["continent"] == cont]["beer_servings"].mean()
    print(cont, res)
```

```python
drinks.groupby("continent").min()
```

```python
drinks.groupby("continent").max()
```

```python
drinks.groupby("continent").mean(numeric_only=True)
```

```python
drinks.groupby("continent")["beer_servings"].mean()
```

```python
drinks.groupby("continent")["beer_servings"].mean().plot(kind="bar")
```

```python
drinks.groupby("continent")["beer_servings"].mean().sort_values().plot(kind="bar")
```

```python
drinks.groupby("continent").mean(numeric_only=True).plot(kind="bar")
```
